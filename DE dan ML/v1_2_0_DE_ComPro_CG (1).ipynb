```python
# ===== IMPORTS =====
import pandas as pd
import numpy as np
import re
import html
import os

base_path = "/content/drive/MyDrive/Semester 7 /Computing project "  # pakai path yang ada spasi di akhir sesuai Drive kamu
# paths
score_path = os.path.join(base_path, "merged_score_data_raw.xlsx")
logs_path  = os.path.join(base_path, "merged_logs_data_raw.xlsx")

# ===== LOAD =====
score_df = pd.read_excel(score_path)
logs_df  = pd.read_excel(logs_path)

# ===== CLEAN SCORE DF =====
# Convert epoch-like columns to datetime when possible
for col in ["quiz_start_time", "quiz_finish_time", "graded_time", "submission_time"]:
    if col in score_df.columns:
        # some values might be like 1727... seconds since epoch; use errors='coerce'
        score_df[col] = pd.to_datetime(score_df[col], unit='s', errors='coerce')

# If human columns exist and are strings, try parse them (fallback)
if "quiz_start_time_human" in score_df.columns:
    score_df["quiz_start_time_human_parsed"] = pd.to_datetime(score_df["quiz_start_time_human"], errors='coerce', dayfirst=True)
if "quiz_finish_time_human" in score_df.columns:
    score_df["quiz_finish_time_human_parsed"] = pd.to_datetime(score_df["quiz_finish_time_human"], errors='coerce', dayfirst=True)

# Score percentage (handle division by zero)
score_df["score_pct"] = np.where(
    score_df["raw_quiz_score"].notna() & score_df["max_quiz_score"].notna(),
    (score_df["raw_quiz_score"] / score_df["max_quiz_score"]) * 100,
    np.nan
)

# Flag attempt
score_df["is_attempt"] = score_df["quiz_state"].fillna("") == "finished"

# Optional: drop completely-empty columns (if >90% null)
null_frac = score_df.isnull().mean()
drop_cols = null_frac[null_frac>0.9].index.tolist()
# don't automatically drop unless you want; show them
print("Kolom dengan >90% null (pertimbangkan drop):", drop_cols)

# ===== CLEAN LOGS DF =====
# 1) parse Time column (format dd/mm/yy, HH:MM:SS)
logs_df["Time_parsed"] = pd.to_datetime(logs_df["Time"], format="%d/%m/%y, %H:%M:%S", errors="coerce", dayfirst=True)

# 2) extract actor_userid from Description by unescape html then regex
def extract_userid_from_desc(txt):
    if pd.isna(txt): return np.nan
    t = html.unescape(str(txt))
    m = re.search(r"id\s+'?(\d+)'?", t)  # matches id '12345' or id 12345
    if m:
        return int(m.group(1))
    # fallback: find first large number
    m2 = re.search(r"(\d{4,})", t)
    if m2:
        return int(m2.group(1))
    return np.nan

logs_df["actor_userid"] = logs_df["Description"].apply(extract_userid_from_desc)

# 3) event context type and detail
logs_df["event_context_type"] = logs_df["Event context"].str.split(":", n=1).str[0].str.strip()
logs_df["event_context_detail"] = logs_df["Event context"].str.split(":", n=1).str[1].str.strip().fillna("")

# 4) simplify event type from Event name
def normalize_event_type(evname):
    ev = str(evname).lower()
    if "view" in ev: return "view"
    if "update" in ev or "edited" in ev or "submitted" in ev: return "update"
    if "report" in ev or "log report" in ev: return "report"
    if "delete" in ev: return "delete"
    return "other"

logs_df["event_type"] = logs_df["Event name"].apply(normalize_event_type)

# 5) actor_userid sometimes absent: fallback to actor from "User full name" if mapping known
# (we will use `actor_userid` as key; but if missing, we will use 'userid' from score later)

# ===== AGGREGATE FEATURES per user from SCORE =====
# basic aggregates for each userid
score_agg = score_df.groupby("userid").agg(
    num_rows = ("userid","size"),
    num_attempts = ("is_attempt","sum"),
    mean_score_pct = ("score_pct","mean"),
    median_score_pct = ("score_pct","median"),
    num_quizzes_taken = ("raw_quiz_score","count"),
    avg_time_spent_seconds = ("time_spent_seconds","mean")
).reset_index()

score_agg["mean_score_pct"] = score_agg["mean_score_pct"].round(2)

# ===== AGGREGATE FEATURES per user from LOGS =====
logs_df["actor_userid"] = logs_df["actor_userid"].astype("Int64")  # allow NA
logs_agg = logs_df.groupby("actor_userid").agg(
    total_events = ("Time_parsed","size"),
    unique_courses = ("event_context_detail", lambda s: s.nunique()),
    first_activity = ("Time_parsed","min"),
    last_activity  = ("Time_parsed","max"),
    num_views = ("event_type", lambda s: (s=="view").sum()),
    num_reports = ("event_type", lambda s: (s=="report").sum())
).reset_index().rename(columns={"actor_userid":"userid"})

# compute active days
logs_agg["active_days"] = (logs_agg["last_activity"] - logs_agg["first_activity"]).dt.days.fillna(0).astype(int)
logs_agg["events_per_day"] = (logs_agg["total_events"] / (logs_agg["active_days"].replace(0,1))).round(2)

# ===== MERGE AGGREGATES =====
user_df = pd.merge(score_agg, logs_agg, on="userid", how="outer")

# Fill NaN numeric with 0 for convenience
num_cols = user_df.select_dtypes(include=["float","int"]).columns.tolist()
user_df[num_cols] = user_df[num_cols].fillna(0)

# ===== SAVE cleaned/aggregated files to Drive =====
out1 = os.path.join(base_path, "merged_score_data_cleaned.csv")
out2 = os.path.join(base_path, "merged_logs_data_cleaned.csv")
out3 = os.path.join(base_path, "user_level_features.csv")

score_df.to_csv(out1, index=False)
logs_df.to_csv(out2, index=False)
user_df.to_csv(out3, index=False)

print("Saved:", out1, out2, out3)
```

```python
# ================= FINAL DATA PREP + CLUSTERING v1 =================
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans

# Load user-level dataset sebelumnya
user_df = pd.read_csv(base_path + "/user_level_features.csv")

# ====== 1️⃣ Filter non-student accounts ======
# Asumsi mahasiswa PJJ memiliki userid >= 10000
clean_df = user_df[user_df["userid"] >= 10000].copy()

print("Before filter:", len(user_df))
print("After filter:", len(clean_df))

# ====== 2️⃣ Feature Engineering ======

# Engagement score = total_events + num_views + active_days (bobot masih equal)
clean_df["engagement_score"] = (
    clean_df["total_events"] +
    clean_df["num_views"] +
    clean_df["active_days"]
)

# Consistency score = events/day
clean_df["consistency_score"] = clean_df["events_per_day"]

# Performance category untuk evaluasi rekomendasi
def performance_cat(x):
    if x >= 80:
        return "High"
    elif x >= 60:
        return "Medium"
    else:
        return "Low"

clean_df["performance_category"] = clean_df["mean_score_pct"].apply(performance_cat)

# ====== 3️⃣ Select Features untuk ML ======
features = clean_df[[
    "mean_score_pct",
    "num_attempts",
    "num_quizzes_taken",
    "engagement_score",
    "consistency_score"
]].copy()

# ====== 4️⃣ Scaling agar clustering tidak berat sebelah ======
scaler = MinMaxScaler()
scaled_features = scaler.fit_transform(features)
scaled_df = pd.DataFrame(scaled_features, columns=features.columns)

# ====== 5️⃣ K-Means Clustering v1 ======
kmeans = KMeans(n_clusters=4, random_state=42)
clean_df["cluster"] = kmeans.fit_predict(scaled_df)

print("Cluster distribution:")
print(clean_df["cluster"].value_counts())

# ====== 6️⃣ Save Final Dataset for ML + Backend ======
final_path = base_path + "/user_level_features_final_for_ML.csv"
clean_df.to_csv(final_path, index=False)
print("Saved final file to:", final_path)

# ====== 7️⃣ Preview ======
print("\n===== Final Preview =====")
display(
    clean_df.head(10)[
        [
            "userid", "mean_score_pct", "num_attempts",
            "engagement_score", "consistency_score",
            "performance_category", "cluster"
        ]
    ]
)
```
